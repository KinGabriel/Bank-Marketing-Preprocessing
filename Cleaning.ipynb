```python
import numpy as np
import pandas as pd
import datetime
```

```python
# Read the CSV file
data = pd.read_csv('bank_marketing.csv')
data
```

```python
# Check if there are duplicated values in the data set
data.duplicated()
```

```python
# Remove the duplicates in the dataset which targets the primary key client_id
data = data.drop_duplicates(subset=['client_id'], keep='first').copy()
data
```

```python
# Define a mapping of month names to numerical values
month_mapping = {
    'jan': 1, 'feb': 2, 'mar': 3, 'apr': 4, 'may': 5, 'jun': 6,
    'jul': 7, 'aug': 8, 'sep': 9, 'oct': 10, 'nov': 11, 'dec': 12
}
```

```python
# Add a temporary year column (2022) that will be used for formatting later
data['year'] = 2022
```

```python
# Replace month names with numeric values
data['month'] = data['month'].map(month_mapping)
data
```

```python
# Combine year, month, and day into a single date column
data['date'] = pd.to_datetime(data[['year', 'month', 'day']])
data
```

```python
# Drop unnecessary and temporary columns
data = data.drop(columns=['month', 'year', 'day'])
data
```

```python
# Convert the values of 'contact_duration' into numeric 
data['contact_duration'] = pd.to_numeric(data['contact_duration'], errors='coerce')
data
```

```python
# Convert seconds to 'HH:MM:SS' format
data['contact_duration'] = data['contact_duration'].apply(lambda x: f'{int(x//3600)}:{int((x%3600)//60):02d}:{int(x%60):02d}')
data

```

```python
# Convert 'campaign_outcome' values from 'yes'/'no' to 'Successful'/'Failed'
data['campaign_outcome'] = data['campaign_outcome'].map({'yes': 'Successful', 'no': 'Failed'})
data
```

```python
# Capitalize first letter of every word in all string columns
data = data.apply(lambda x: x.str.title() if x.dtype == "object" else x)
data
```

```python
# Replace "Unknown" values with NaN
data = data.replace("Unknown", np.nan)
data
```

```python
#Impute age column with median value, so even if there are outliers, the imputed values are not affect
data['age'] = data['age'].fillna(data['age'].median())
data
```

```python
#Impute job column with the most frequent value (mode)
data['job'] = data['job'].fillna(data['job'].mode()[0])
data
```

```python
#Impute credit_default column with the most frequent value (mode)
data['credit_default'] = data['credit_default'].fillna(data['credit_default'].mode()[0])
data
```

```python
#Impute mortgage column with the most frequent value (mode)
data['mortgage'] = data['mortgage'].fillna(data['mortgage'].mode()[0])
data
```

```python
#Imputation for education; based on most common education for each job
most_common_education_per_job = data.groupby('job')['education'].agg(lambda x: x.mode()[0])
#Map the most common education per job based on the groupby method
data.loc[data['education'].isnull(), 'education'] = data.loc[data['education'].isnull(), 'job'].map(most_common_education_per_job)
data
```

```python
data.to_csv('bank_marketing_clean.csv', index=False)
```
